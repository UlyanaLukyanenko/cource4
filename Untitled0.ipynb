```julia
function print_sin_table(increment, max_angle)
    angle = 0
    while angle <= max_angle
        rad = deg2rad(angle)
        x = sin(rad)
        println(x)
        angle = angle + increment
    end
end
```

```julia
function sine(x)
    n = 5
    total = 0
    for i in 0:n
        total += (-1)^i*x^(2i+1)/factorial(2i + 1)
    end
    total
end
```

```julia
function fac(n)
    if n > 2
        n*fac(n-1)
    elseif n > 0
        n
    elseif n == 0
        1
    else
        err = DomainError(n, "`n` must not be negative.")
        throw(err)
    end
end
```

```julia
function fib(n)
    if n == 0
        0
    elseif n == 1
        1
    else
        fib(n-1) + fib(n-2)
    end
end
```

```julia
function fibi(n)
    if 0 <= n <= 1
        return n
    end
    prev = 0
    x = 1
    for i in 2:n
        tmp = x
        x += prev
        prev = tmp
    end
    x
end
```

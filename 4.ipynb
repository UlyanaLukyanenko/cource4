```julia
function print_sin_table(increment, max_angle)
    angle = 0
    while angle <= max_angle
        rad = deg2rad(angle)
        x = sin(rad)
        println(x)
        angle = angle + increment
    end
end
```

```julia
function sine(x)
    n = 5
    total = 0
    for i in 0:n
        total += (-1)^i*x^(2i+1)/factorial(2i + 1)
    end
    total
end
```

```julia
function sinus(x)
    n = 5
    taylor(i) = (-1)^i*x^(2i+1)/factorial(2i + 1)
    mapreduce(taylor, +, 0:n)
end
```
